# Generar Cartillas para TODAS las Estaciones NO Problemáticas

Este notebook genera automáticamente las 5 figuras técnicas para todas las estaciones EXCEPTO las 5 problemáticas identificadas.

## Estaciones a Generar (13 estaciones):

**Zona Baja (6):** CALAMAR, BARBOSA, TENERIFE, PLATO, MAGANGUE, SAN_PEDRITO

**Zona Media (3):** LAS_AGUADAS, EL_BANCO, PENONCITO

**Zona Alta (4):** LA_GLORIA, EL_CONTENTO, SITIO_NUEVO, PUERTO_WILCHES

---

## Instrucciones:
1. Ejecutar la celda de abajo (Cell → Run All)
2. Esperar ~10-15 minutos (genera 13 estaciones × 5 figuras = 65 imágenes)
3. Revisar la carpeta `resultados/` con subcarpetas por estación

````python
# Lista de estaciones NO problemáticas
ESTACIONES_NO_PROBLEMATICAS = [
    # Zona Baja (6 estaciones)
    'CALAMAR', 'BARBOSA', 'TENERIFE', 'PLATO', 'MAGANGUE', 'SAN_PEDRITO',
    
    # Zona Media (3 estaciones - excluyendo RIO_NUEVO, LAS_PALOMAS, ARMENIA)
    'LAS_AGUADAS', 'EL_BANCO', 'PENONCITO',
    
    # Zona Alta (4 estaciones - excluyendo GAMARRA_AUT)
    'LA_GLORIA', 'EL_CONTENTO', 'SITIO_NUEVO', 'PUERTO_WILCHES'
]

print(f"Estaciones a procesar: {len(ESTACIONES_NO_PROBLEMATICAS)}")
print("\nLista de estaciones:")
for i, est in enumerate(ESTACIONES_NO_PROBLEMATICAS, 1):
    zona = "Baja" if i <= 6 else "Media" if i <= 9 else "Alta"
    print(f"  {i:2d}. {est:20s} (Zona {zona})")
````

````python
# Importar funciones
import sys
import os
sys.path.append('./funciones')
from graficador import generar_cartilla_completa

print("✓ Funciones importadas")
````

````python
# Generar todas las cartillas en subcarpetas

exitosos = 0
fallidos = 0

for i, estacion in enumerate(ESTACIONES_NO_PROBLEMATICAS, 1):
    # Crear subcarpeta
    estacion_dir = f"./resultados/{estacion}"
    os.makedirs(estacion_dir, exist_ok=True)
    
    print(f"\n[{i}/{len(ESTACIONES_NO_PROBLEMATICAS)}] {estacion}")
    print("-" * 50)
    
    try:
        generar_cartilla_completa(estacion, output_dir=estacion_dir + '/')
        exitosos += 1
    except Exception as e:
        print(f"✗ Error: {e}")
        fallidos += 1

print("\n" + "="*60)
print(f"COMPLETADO: {exitosos} exitosos, {fallidos} fallidos")
print("="*60)
````
